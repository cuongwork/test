3.1 Chuẩn hóa dữ liệu

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

def analyze_skewness(x):
    fig, ax = plt.subplots(2, 2, figsize=(10, 8))  # Tăng kích thước biểu đồ

    # Chuyển đổi dữ liệu để tránh lỗi
    log_data = np.log1p(rfmTable[x])  # Sử dụng log1p để tránh log(0)
    sqrt_data = np.sqrt(rfmTable[x])  # Căn bậc hai
    boxcox_data, _ = stats.boxcox(rfmTable[x] + 1)  # Cộng thêm 1 để tránh lỗi với giá trị 0

    # Biểu đồ phân phối gốc
    sns.histplot(rfmTable[x], bins=30, kde=True, ax=ax[0,0])
    ax[0,0].set_title(f'Original Distribution of {x}')

    # Biểu đồ phân phối sau khi log-transform
    sns.histplot(log_data, bins=30, kde=True, ax=ax[0,1])
    ax[0,1].set_title(f'Log Transformed Distribution of {x}')

    # Biểu đồ phân phối sau khi sqrt-transform
    sns.histplot(sqrt_data, bins=30, kde=True, ax=ax[1,0])
    ax[1,0].set_title(f'Square Root Transformed Distribution of {x}')

    # Biểu đồ phân phối sau khi Box-Cox transform
    sns.histplot(boxcox_data, bins=30, kde=True, ax=ax[1,1])
    ax[1,1].set_title(f'Box-Cox Transformed Distribution of {x}')

    plt.tight_layout()
    plt.show()

    # In giá trị skewness của từng phương pháp
    print(f"Original Skewness: {rfmTable[x].skew().round(2)}")
    print(f"Log Transformed Skewness: {log_data.skew().round(2)}")
    print(f"Square Root Transformed Skewness: {sqrt_data.skew().round(2)}")
    print(f"Box-Cox Transformed Skewness: {pd.Series(boxcox_data).skew().round(2)}")
```

3.1.1 Độ lệch qua các phép biến đổi với R




```python
analyze_skewness('Recency')
```

3.1.2 Độ lệch qua các phép biến đổi với F


```python
analyze_skewness('Frequency')
```

3.1.3 Độ lệch qua các phép biến đổi với M


```python
analyze_skewness('Monetary_Value')
```

3.2 Sử dụng Box-Cox để chuẩn hóa

```python
# Tạo DataFrame mới để lưu dữ liệu đã biến đổi
df_customers_t = pd.DataFrame()

# Cộng thêm 1 vào Recency để tránh lỗi với giá trị 0 (Box-Cox yêu cầu dữ liệu dương)
df_customers_t['Recency'] = stats.boxcox(rfmTable['Recency'] + 1)[0]

# Frequency luôn dương, vì vậy có thể áp dụng trực tiếp Box-Cox mà không cần điều chỉnh
df_customers_t['Frequency'] = stats.boxcox(rfmTable['Frequency'])[0]

# Dùng Box-Cox cho Monetary_Value để giảm độ lệch phân phối.
df_customers_t['Monetary_Value'] = stats.boxcox(rfmTable['Monetary_Value'])[0]

df_customers_t
```

```python
df_customers_t.describe()
```

```python
# Tien hanh scale du lieu

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(df_customers_t)

df_customers_t = scaler.transform(df_customers_t)
```

3.3 Tìm số cụm với Elbow

```python
# Chọn số cụm bằng Elbow
from sklearn.cluster import KMeans

sse = {}
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(df_customers_t)
    sse[k] = kmeans.inertia_

plt.title('The Elbow Method')
plt.xlabel('k')
plt.ylabel('SSE')
sns.pointplot(x=list(sse.keys()), y=list(sse.values()))
plt.show()
```

3.4 Kiểm định chất lượng cụm với thuật toán Silhoutte


```python
from sklearn.metrics import silhouette_samples, silhouette_score
# find the optimum number of clusters (k) using the Silhouette method
for n_clusters in range(2,8):
    kmeans = KMeans(init='k-means++', n_clusters = n_clusters, n_init=100)
    kmeans.fit(df_customers_t)
    clusters = kmeans.predict(df_customers_t)
    silhouette_avg = silhouette_score(df_customers_t, clusters)

    print("For n_clusters =", n_clusters, "The average silhouette_score is :", silhouette_avg)
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from yellowbrick.cluster import SilhouetteVisualizer

# Dữ liệu của bạn: df_customers_t (giả sử đã chuẩn hóa, chỉ gồm dữ liệu số)

range_n_clusters = range(2, 6)

for n_clusters in range_n_clusters:
    print(f"\n🔹 Đang kiểm tra với n_clusters = {n_clusters}...")

    # Khởi tạo mô hình KMeans
    model = KMeans(init='k-means++', n_clusters=n_clusters, n_init=100, random_state=0)

    # Fit mô hình để lấy nhãn cụm
    clusters = model.fit_predict(df_customers_t)

    # Tính điểm silhouette trung bình
    silhouette_avg = silhouette_score(df_customers_t, clusters)
    print(f"👉 Silhouette Score trung bình: {silhouette_avg:.4f}")

    # Tạo biểu đồ silhouette
    visualizer = SilhouetteVisualizer(model, colors='yellowbrick')
    visualizer.fit(df_customers_t)
    visualizer.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from yellowbrick.cluster import SilhouetteVisualizer

# Dữ liệu của bạn: df_customers_t (giả sử đã chuẩn hóa, chỉ gồm dữ liệu số)

range_n_clusters = range(2, 6)

# Tạo figure với 2 hàng 2 cột
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.flatten()  # Chuyển thành mảng 1 chiều để dễ index

for idx, n_clusters in enumerate(range_n_clusters):
    print(f"\n🔹 Đang kiểm tra với n_clusters = {n_clusters}...")

    # Khởi tạo mô hình KMeans
    model = KMeans(init='k-means++', n_clusters=n_clusters, n_init=100, random_state=0)

    # Fit mô hình để lấy nhãn cụm
    clusters = model.fit_predict(df_customers_t)

    # Tính điểm silhouette trung bình
    silhouette_avg = silhouette_score(df_customers_t, clusters)
    print(f"👉 Silhouette Score trung bình: {silhouette_avg:.4f}")

    # Vẽ biểu đồ silhouette lên subplot tương ứng
    visualizer = SilhouetteVisualizer(model, colors='yellowbrick', ax=axes[idx])
    visualizer.fit(df_customers_t)

# Căn chỉnh layout đẹp hơn
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Sao chép dữ liệu đã phân cụm (đừng fit lại mô hình!)
KM5_clustered = rfmTable.copy()

# Vẽ biểu đồ 3D
fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(111, projection='3d', elev=20, azim=210)

# Lấy danh sách các cụm
clusters = KM5_clustered['K_Cluster'].unique()

# Ánh xạ màu thủ công để đảm bảo màu trùng với cmap=Set1
colors = plt.cm.Set1(np.linspace(0, 1, len(clusters)))

# Vẽ từng cụm riêng để có legend
for cluster_id, color in zip(clusters, colors):
    cluster_data = KM5_clustered[KM5_clustered['K_Cluster'] == cluster_id]
    ax.scatter(
        cluster_data['Recency'],
        cluster_data['Frequency'],
        cluster_data['Monetary_Value'],
        c=[color],
        label=f'cum {cluster_id}',
        s=35
    )

# Tùy chỉnh hiển thị
ax.set_xlabel('Recency')
ax.set_ylabel('Frequency')
ax.set_zlabel('Monetary_Value')
ax.set_title('Che đo xem 3D của 4 cum K-Means')

# Ẩn nhãn trục nếu muốn biểu đồ sạch hơn
ax.tick_params(axis='x', labelbottom=False)
ax.tick_params(axis='y', labelleft=False)
ax.tick_params(axis='z', labelleft=False)

# Thêm chú thích
ax.legend(loc='upper left')

plt.show()
```
